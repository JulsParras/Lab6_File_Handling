```python
LAST_NAME = "PARRAS"
SEED_NUM = 7
FAVORITE_ARTIST = "DAUGHTRY"
```

```python
def generate_readings():
    readings = [
        len(LAST_NAME),
        SEED_NUM,
        len(FAVORITE_ARTIST)
    ]
    return readings

    
```

```python
def validate_readings(readings):
    for reading in readings:
        if reading < 0:
            raise ValueError("Invalid reading.")
    return True
    
```

```python
def calculate_average(readings):
    total = sum(readings)
    average = total / len(readings)
    return average

    
```

```python
def classify_equipment(average):
    if average >= 10:
        return "NORMAL"
    else:
        return "CHECK"
        
```

```python
def diagnostic_log(func):
    def wrapper(*args):
        print("Running:", func.__name__)
        result = func(*args)
        print("Finished:", func.__name__)
        return result
    return wrapper
    
```

```python
@diagnostic_log
def diagnose(readings):
    validate_readings(readings)
    average = calculate_average(readings)
    condition = classify_equipment(average)
    
    return average, condition

try:
    readings = generate_readings()

    average, condition = diagnose(readings)

    print("Generated Equipment Data:", readings)
    print("Average:", round(average, 2))
    print("Diagnostic Result:", condition)

except ValueError as e:
    print("Validation Error:", e)
    
```
