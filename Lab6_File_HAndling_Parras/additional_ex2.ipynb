```python

# Import Python tools for reading, writing, and backing up files.
import csv
import json
import shutil
from pathlib import Path

# Set the location of your dataset.
input_file = Path(r"C:\Users\JULIANNA\Downloads\user_behavior_dataset.csv")


# Generate the required operating system.
if SEED_DIGIT % 2 == 0:
    target_os = "Android"
else:
    target_os = "iOS"

# Generate the required user behavior class.
target_class = (SEED_DIGIT % 5) + 1

# Set the names of the output files.
filtered_file = Path.cwd() / "filtered_usage.csv"
summary_file = Path.cwd() / "usage_summary.json"
backup_file = Path.cwd() / "filtered_usage_backup.csv"

# Display the student-specific criteria.
print("Operating system:", target_os)
print("Target behavior class:", target_class)



# Read all records from the CSV file.
with open(input_file, "r", newline="", encoding="utf-8-sig") as file:
    reader = csv.DictReader(file)
    data = list(reader)

# Display the number of records loaded.
print("Total records in dataset:", len(data))
```

```python

# Prepare a list for the selected users.
selected_records = []



# Check each record in the dataset.
for row in data:

    # Read the operating system and behavior class.
    operating_system = row["Operating System"].strip()
    behavior_class = int(row["User Behavior Class"])

    # Select records matching BOTH assigned criteria.
    if operating_system == target_os and behavior_class == target_class:
        selected_records.append(row)

# Count the selected records.
selected_count = len(selected_records)

# Calculate averages only when matching records exist.
if selected_count > 0:

    # Add the battery drain values and divide by the number of users.
    average_battery = sum(
        float(row["Battery Drain (mAh/day)"])
        for row in selected_records
    ) / selected_count

    # Add the data usage values and divide by the number of users.
    average_data = sum(
        float(row["Data Usage (MB/day)"])
        for row in selected_records
    ) / selected_count

else:
    # Use None when no records match, instead of dividing by zero.
    average_battery = None
    average_data = None

# Display the results.
print("Records matching operating system:", 
      sum(1 for row in data if row["Operating System"].strip() == target_os))

print("Records matching target behavior class:", selected_count)
print("Average battery drain:", average_battery)
print("Average data usage:", average_data)
```

```python
# Check whether the dataset variable exists.
print("data" in globals())
```

```python
# Define the exact columns required in the output CSV.
export_fields = [
    "User ID",
    "Device Model",
    "Operating System",
    "Battery Drain (mAh/day)",
    "Data Usage (MB/day)"
]

# Open a new CSV file for writing.
with open(filtered_file, "w", newline="", encoding="utf-8") as file:

    # Write only the five required columns.
    writer = csv.DictWriter(
        file,
        fieldnames=export_fields,
        extrasaction="ignore"
    )

    # Write the column headers.
    writer.writeheader()

    # Export only the selected users.
    writer.writerows(selected_records)

# Display the export results.
print("Exported CSV file:", filtered_file)
print("Records exported:", len(selected_records))
```

```python

# Prepare the summary using the calculated results.
summary = {
    "operating_system": target_os,
    "target_behavior_class": target_class,
    "total_records_in_dataset": len(data),
    "records_matching_operating_system": sum(
        1 for row in data
        if row["Operating System"].strip() == target_os
    ),
    "records_in_target_class": selected_count,
    "average_battery_drain_mAh_per_day": average_battery,
    "average_data_usage_MB_per_day": average_data
}

# Save the summary as a JSON file.
with open(summary_file, "w", encoding="utf-8") as file:
    json.dump(summary, file, indent=4)

# Read the exported CSV to count its actual data rows.
with open(filtered_file, "r", newline="", encoding="utf-8") as file:
    exported_count = sum(1 for row in csv.DictReader(file))

# Verify that the exported row count matches the selected count.
count_verified = exported_count == selected_count

# Display the summary and verification result.
print("JSON summary file:", summary_file)
print(json.dumps(summary, indent=4))
print("Expected exported rows:", selected_count)
print("Actual exported rows:", exported_count)
print("Count verification:", count_verified)
```

```python

# Copy the exported CSV into a backup file.
shutil.copy2(filtered_file, backup_file)

# Get the size of the backup file in bytes.
backup_size = backup_file.stat().st_size

# Verify that the backup and exported CSV have the same size.
backup_verified = backup_size == filtered_file.stat().st_size

# Display the backup information.
print("Backup file:", backup_file)
print("Backup file size:", backup_size, "bytes")
print("Backup verification:", backup_verified)
```
